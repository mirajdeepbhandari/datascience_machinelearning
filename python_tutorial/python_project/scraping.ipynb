### Disable SSL 

```python
import requests
import urllib3

def disable_ssl_verification():
    """Disable SSL certificate verification globally for requests."""
    urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

    original_request = requests.Session.request

    def request_without_ssl(self, *args, **kwargs):
        kwargs["verify"] = False
        return original_request(self, *args, **kwargs)

    requests.Session.request = request_without_ssl
    
disable_ssl_verification()
```

## Getting the HTML Page

```python
import requests

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36",
    "Accept-Language": "en-US,en;q=0.9",
}

response = requests.get("https://yamahamusicstore.in/", headers=headers, timeout=10)

print(response.status_code)
```

```python
html_content=response.text
html_content
```

## Converting the HTML String to Beautiful Soup Object

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(html_content, 'html.parser')
print(soup.prettify())
```

```python
category_section = soup.find('div', class_='tab-content trandingContainer')
print(category_section.prettify())
```

```python
product_cards = soup.find_all('div', class_='catgcvr')
product_cards
```

## Make Category and Link Dict

```python
categories = {}
for a_tag in product_cards:
    category_name = a_tag.get_text(strip=True)
    link = a_tag.find('a').get('href')
    categories[category_name] = link

categories
```

##  Lets Go inside one page (Keyboard Instruments) and scrape the content 

```python
html_response = requests.get(categories['Keyboard Instruments'], headers=headers, timeout=10)
html_response.status_code
if html_response.status_code != 200:
    print(f"Failed to fetch the page")
else:
    keyboard_html = html_response.text


keyboard_html
```

```python
## convert keyboard html to soup
keyboard_soup = BeautifulSoup(keyboard_html, 'html.parser')
print(keyboard_soup.prettify())
```

```python
keyboard_products_section = keyboard_soup.find("div", class_="tf-list-layout wrapper-shop prodLising")
print(keyboard_products_section)
```

## Now Looping over all the Products and Getting the Information

```python
product_info = {'Keyboard Instruments':[]}

for soup_item in actual_card:
    title = soup_item.find("a").get_text(strip=True)
    rating = soup_item.find("div", class_="list-star").get_text(strip=True)
    price = soup_item.find("div", class_="price").get_text(strip=True)
    link = soup_item.find("a").get('href')
    
    product_info['Keyboard Instruments'].append({
        "title": title,
        "rating": rating,
        "price": price,
        "link": link
    })
    
```

## now to get the data from another Page also 

### www.examplelink.com`?page=n`

```python
total_product_info ={"Keyboard Instruments": []} 

total_page_needed = 10

for page in range(1, total_page_needed + 1):
    print(f"Scraping page {page} of {total_page_needed}...\n")
    page_url = f"{categories['Keyboard Instruments']}?page={page}"
    html_response = requests.get(page_url, headers=headers, timeout=10)
    
    if html_response.status_code != 200:
        print(f"Failed to fetch page {page}")
        break
    
    keyboard_html = html_response.text
    keyboard_soup = BeautifulSoup(keyboard_html, 'html.parser')
    keyboard_products_section = keyboard_soup.find("div", class_="tf-list-layout wrapper-shop prodLising")
    
    if not keyboard_products_section:
        print(f"No products found on page {page}")
        break
    
    actual_card = keyboard_products_section.find_all("div", class_="card-product-info")
    
    for soup_item in actual_card:
        title = soup_item.find("a").get_text(strip=True)
        rating = soup_item.find("div", class_="list-star").get_text(strip=True)
        price = soup_item.find("div", class_="price").get_text(strip=True)
        link = soup_item.find("a").get('href')
        
        total_product_info['Keyboard Instruments'].append({
            "title": title,
            "rating": rating,
            "price": price,
            "link": link
        })  
        

total_product_info
```

###  So We have one Scrapped for one Category Keyboard lets do for all Category now

```python
categories
```

```python
all_products_info={category: [] for category in categories.keys()} 
all_products_info
```

```python
total_page_needed = 10

for category_name, category_link in categories.items():

    print(f"\n===== {category_name} =====")

    for page in range(1, total_page_needed + 1):

        print(f"Scraping page {page}...")

        page_url = f"{category_link}?page={page}"
        html_response = requests.get(page_url, headers=headers, timeout=10)

        if html_response.status_code != 200:
            print(f"Failed to fetch page {page}")
            break

        category_soup = BeautifulSoup(html_response.text, "html.parser")

        category_products_section = category_soup.find(
            "div",
            class_="tf-list-layout wrapper-shop prodLising"
        )

        if category_products_section is None:
            print(f"No more products found after page {page - 1}")
            break

        actual_card = category_products_section.find_all(
            "div",
            class_="card-product-info"
        )

        if not actual_card:
            print(f"No products found on page {page}")
            break

        print(f"Found {len(actual_card)} products")

        for soup_item in actual_card:

            title_tag = soup_item.find("a")
            rating_tag = soup_item.find("div", class_="list-star")
            price_tag = soup_item.find("div", class_="price")
            title = title_tag.get_text(strip=True) if title_tag else None
            rating = rating_tag.get_text(strip=True) if rating_tag else None
            price = price_tag.get_text(strip=True) if price_tag else None
            link = title_tag.get("href") if title_tag else None

            all_products_info[category_name].append({
                "title": title,
                "rating": rating,
                "price": price,
                "link": link
            })

print("\nScraping completed for all categories.")
```

```python
total_all_products = sum(len(products) for products in all_products_info.values())
total_all_products
```
