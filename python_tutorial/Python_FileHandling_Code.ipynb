# 1. Text File Operations

## 1.1 Why File Handling?

Programs often need to **save data permanently** or **load existing data**:

- Save user settings, logs, reports.
- Read data files (text, CSV, JSON, etc.).
- Process large datasets that don't fit in memory.

Python uses the built-in **`open()`** function to work with files. The general workflow is:

```text
open() -> read/write -> close()
```

### Syntax

```python
file = open("path/to/file.txt", "mode", encoding="utf-8")
# ... do stuff ...
file.close()
```

> **Best practice:** Always use the `with` statement (covered below). It auto-closes the file even if an error occurs.

## 1.2 File Modes

The **mode** tells Python what you want to do with the file.

| Mode | Meaning | File must exist? | Overwrites? |
|------|---------|------------------|-------------|
| `r`  | Read (text) - default | Yes | No |
| `w`  | Write (text) - creates/overwrites | No | YES, erases content |
| `a`  | Append (text) - adds at the end | No | No |
| `x`  | Exclusive create - fails if file exists | No | No |
| `r+` | Read & write | Yes | No |
| `rb` | Read in **binary** mode | Yes | No |
| `wb` | Write in **binary** mode | No | YES |
| `ab` | Append in **binary** mode | No | No |

**Text vs Binary:**
- Text mode (`r`, `w`) - handles strings (`str`), uses encoding (default UTF-8).
- Binary mode (`rb`, `wb`) - handles raw bytes (`bytes`). Used for images, PDFs, audio, etc.

````python
# Writing to a text file with mode 'w' (creates / overwrites)
file = open("sample.txt", "w", encoding="utf-8")
file.write("Hello, Python!\n")
file.write("This is line 2.\n")
file.write("This is line 3.\n")
file.close()                          # always close after use

print("File 'sample.txt' created successfully.")
````

## 1.3 The `with` Statement (Recommended)

Using `with` automatically closes the file for you - even if an error happens in the block.

```python
with open("file.txt", "r") as f:
    data = f.read()
# file is now closed automatically
```

````python
# Recommended way - using 'with'
with open("sample.txt", "r", encoding="utf-8") as f:
    content = f.read()

print(content)
print("File closed?", f.closed)        # True - auto-closed
````

## 1.4 Reading Methods

| Method | What it does |
|--------|--------------|
| `read()` | Reads the entire file as a single string |
| `read(n)` | Reads the first `n` characters |
| `readline()` | Reads one line at a time |
| `readlines()` | Reads all lines into a list |
| `for line in file:` | Memory-efficient line-by-line loop |

````python
# read() - whole file as a single string
with open("sample.txt", "r") as f:
    data = f.read()
print("read():")
print(data)

# read(n) - read only n characters
with open("sample.txt", "r") as f:
    print("read(5):", repr(f.read(5)))
````

````python
# readline() - one line at a time
with open("sample.txt", "r") as f:
    print("Line 1:", f.readline().strip())
    print("Line 2:", f.readline().strip())

# readlines() - all lines as a list
with open("sample.txt", "r") as f:
    lines = f.readlines()
print("\nLines list:", lines)
````

````python
# BEST way for big files - loop line by line
with open("sample.txt", "r") as f:
    for _, line in enumerate(f, start=1):
        print(line)
````

````python
data = ""
with open("sample.txt", "r") as f:
    for _ , line in enumerate(f, start=1):
        data=data+line
        
print(data)
````

## 1.5 Writing Methods

| Method | What it does |
|--------|--------------|
| `write(text)` | Writes a string to the file |
| `writelines(list)` | Writes a list of strings (no auto-newline) |

Modes:
- `w` -> **overwrites** the file (or creates it).
- `a` -> **appends** to the end (or creates it).

````python
# write() - overwrite mode
with open("notes.txt", "w", encoding="utf-8") as f:
    f.write("First line\n")
    f.write("Second line\n")
````

````python
# write() - overwrite mode
with open("notes.txt", "w", encoding="utf-8") as f:
    f.write("First line\n")
    f.write("Second line\n")

# writelines() - write multiple lines at once (you must add \n yourself)
more_lines = ["Third line\n", "Fourth line\n", "Fifth line\n"]
with open("notes.txt", "a") as f:           # APPEND mode
    f.writelines(more_lines)

# Read it back
with open("notes.txt") as f:
    print(f.read())
````

````python
# Comparing 'w' (overwrite) vs 'a' (append)
with open("demo.txt", "w") as f:
    f.write("Original line\n")

# Append: keeps existing content
with open("demo.txt", "a") as f:
    f.write("Appended line\n")

with open("demo.txt") as f:
    print("--- After append ---")
    print(f.read())

# Overwrite: wipes everything
with open("demo.txt", "w") as f:
    f.write("This replaced everything!\n")

with open("demo.txt") as f:
    print("--- After overwrite ---")
    print(f.read())
````

## 1.6 Binary Mode (`rb`, `wb`)

Use binary mode for **non-text** files: images, PDFs, audio, videos, etc. Data is read/written as `bytes` instead of `str`.

````python
# Writing bytes with 'wb'
data = b"\x48\x65\x6c\x6c\x6f"      # bytes for 'Hello'
with open("binary_demo.bin", "wb") as f:
    f.write(data)

# Reading bytes with 'rb'
with open("binary_demo.bin", "rb") as f:
    raw = f.read()

print("Raw bytes :", raw)
print("As text   :", raw.decode("utf-8"))
print("Type      :", type(raw))
````

````python
# Practical example: copy any file by reading bytes
with open("sample.txt", "rb") as src:
    data = src.read()

with open("sample_copy.txt", "wb") as dst:
    dst.write(data)

print("File copied as sample_copy.txt")
````

## 1.7 File Path Handling with `os` and `os.path`

The **`os` module** provides functions to work with the filesystem in a way that works on Windows, Mac, and Linux.

| Function | Purpose |
|----------|---------|
| `os.getcwd()` | Get the current working directory |
| `os.listdir(path)` | List files & folders in a directory |
| `os.mkdir(path)` | Create a directory |
| `os.makedirs(path)` | Create directories recursively |
| `os.remove(path)` | Delete a file |
| `os.rmdir(path)` | Delete an empty directory |
| `os.rename(old, new)` | Rename a file/folder |
| `os.path.exists(path)` | Check if a path exists |
| `os.path.isfile(path)` | True if path is a file |
| `os.path.isdir(path)` | True if path is a directory |
| `os.path.join(a, b)` | Join path components correctly |
| `os.path.basename(path)` | File name from a path |
| `os.path.dirname(path)` | Folder name from a path |
| `os.path.splitext(path)` | Split into (name, extension) |
| `os.path.getsize(path)` | File size in bytes |

````python
import os

# Current working directory
print("CWD :", os.getcwd())

# List items in the current folder
print("\nItems in CWD:")
for item in os.listdir("."):
    print(" -", item)
````

````python
import os

# Building paths safely with os.path.join() (works on all OSes)
folder = "data"
filename = "sample.txt"
full_path = os.path.join(folder, filename)
print("Joined path:", full_path)

# Splitting paths
path = "C:/Users/Alice/Documents/report.pdf"
print("Base name :", os.path.basename(path))
print("Dir name  :", os.path.dirname(path))
name, ext = os.path.splitext(path)
print("Name      :", name)
print("Extension :", ext)
````

````python
import os

# Existence checks
print("sample.txt exists?", os.path.exists("sample.txt"))
print("Is it a file? ", os.path.isfile("sample.txt"))
print("Is it a dir?  ", os.path.isdir("sample.txt"))

# File size
if os.path.exists("sample.txt"):
    print("Size (bytes):", os.path.getsize("sample.txt"))
````

# 3. Working with JSON

## 3.1 What is JSON?

**JSON** = **JavaScript Object Notation**. It's a lightweight, **language-independent** format for exchanging data.

JSON is everywhere:
- Web APIs (REST responses)
- Configuration files (`.json`)
- NoSQL databases (MongoDB, Firebase)

### JSON Structure (similar to Python)

| JSON | Python Equivalent |
|------|-------------------|
| `{}` object | `dict` |
| `[]` array | `list` |
| `"text"` string | `str` |
| `123` number | `int` / `float` |
| `true` / `false` | `True` / `False` |
| `null` | `None` |

### Example JSON

```json
{
  "name": "Alice",
  "age": 25,
  "is_student": false,
  "hobbies": ["reading", "hiking"],
  "address": {
    "city": "Mumbai",
    "pincode": "400001"
  }
}
```

### The `json` module - 4 main functions

| Function | Purpose |
|----------|---------|
| `json.dumps(obj)` | Convert Python object -> **JSON string** |
| `json.loads(text)` | Convert **JSON string** -> Python object |
| `json.dump(obj, file)` | Write Python object -> **JSON file** |
| `json.load(file)` | Read **JSON file** -> Python object |

> **Memory aid:** functions with `s` (`dumps`/`loads`) deal with **strings**, without `s` deal with **files**.

## 3.2 Converting Between Python and JSON Strings

````python
import json

# Python dict -> JSON string
person = {
    "name": "Alice",
    "age": 25,
    "is_student": False,
    "hobbies": ["reading", "hiking"],
    "address": None
}

json_text = json.dumps(person)
print("JSON string:", json_text)
print("Type       :", type(json_text))
````

````python
import json

person = {
    "name": "Alice",
    "age": 25,
    "hobbies": ["reading", "hiking"],
}

# Pretty-printed JSON with indentation & sorted keys
pretty = json.dumps(person, indent=1, sort_keys=True)
print(pretty)
````

````python
import json

# JSON string -> Python dict (json.loads)
json_text = '{"name": "Bob", "age": 30, "is_admin": true, "tags": ["py", "js"]}'

data = json.loads(json_text)
print("Parsed   :", data)
print("Type     :", type(data))
print("Name     :", data["name"])
print("Is admin :", data["is_admin"])     # JSON 'true' -> Python True
print("Tags     :", data["tags"])
````

````python
import json

# Parsing a JSON ARRAY
json_array = '[1, 2, 3, 4, 5]'
numbers = json.loads(json_array)
print("List :", numbers, "->", type(numbers))

# Parsing a nested JSON
nested = '''
{
    "company": "TechCorp",
    "employees": [
        {"name": "Alice", "role": "Engineer"},
        {"name": "Bob",   "role": "Designer"}
    ]
}
'''
data = json.loads(nested)
print("\nCompany   :", data["company"])
print("First emp :", data["employees"][0]["name"])

for emp in data["employees"]:
    print(f" - {emp['name']} ({emp['role']})")
````

## 3.3 Reading & Writing JSON Files

````python
import json

# Writing Python data to a JSON file with json.dump
student = {
    "name": "Alice",
    "age": 21,
    "subjects": ["Math", "Science", "English"],
    "is_passed": True
}

with open("student.json", "w") as f:
    json.dump(student, f, indent=4)
````

````python
import json

# Writing Python data to a JSON file with json.dump
student = {
    "name": "Alice",
    "age": 21,
    "subjects": ["Math", "Science", "English"],
    "is_passed": True
}

with open("student.json", "w", encoding="utf-8") as f:
    json.dump(student, f, indent=4)

# Verify
with open("student.json") as f:
    print(f.read())
````

````python
import json

# Reading JSON file with json.load
with open("student.json", "r", encoding="utf-8") as f:
    data = json.load(f)

print("Type    :", type(data))
print("Name    :", data["name"])
print("Subjects:", data["subjects"])
print("Passed  :", data["is_passed"])
````

````python
import json

# A list of dicts is a common JSON shape (like a database table)
users = [
    {"id": 1, "name": "Alice",   "email": "alice@example.com"},
    {"id": 2, "name": "Bob",     "email": "bob@example.com"},
    {"id": 3, "name": "Charlie", "email": "charlie@example.com"},
]

with open("users.json", "w") as f:
    json.dump(users, f, indent=2)

# Read it back and process
with open("users.json") as f:
    loaded = json.load(f)

print("Total users:", len(loaded))
for u in loaded:
    print(f" {u['id']}. {u['name']:<8} <{u['email']}>")
````

````python
import json

# Updating a JSON file safely - load, modify, save
with open("users.json") as f:
    users = json.load(f)

# Add a new user
users.append({"id": 4, "name": "Dan", "email": "dan@example.com"})

# Save back
with open("users.json", "w") as f:
    json.dump(users, f, indent=2)

# Verify
with open("users.json") as f:
    print(f.read())
````
