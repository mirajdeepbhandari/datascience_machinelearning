## Accuracy Does Not Always Reflect the Real Performance of a Model

Let's say we have **100 students**:

- **90 students passed**
- **10 students failed**

Now suppose our model predicts that **every student passed**.

### Actual Results

| Actual | Count |
|----------|--------|
| Passed | 90 |
| Failed | 10 |

### Model Predictions

| Predicted | Count |
|------------|--------|
| Passed | 100 |
| Failed | 0 |

### Accuracy Calculation

Accuracy is calculated as:

Accuracy = Correct Predictions / Total Predictions

In this case:

Accuracy = 90 / 100 = 90%

So the model achieves **90% accuracy**.

### The Problem

Even though the accuracy is **90%**, the model completely fails to identify the **10 students who actually failed**.

This shows that a high accuracy score does not necessarily mean the model is performing well, especially when the dataset is **imbalanced**.

Because of this, we also use metrics such as:

- **Precision**
- **Recall**
- **F1 Score**

These metrics provide a more complete picture of the model's performance.

## Precision, Recall, and F1 Score

Continuing with the same example:

- 100 students in total
- 90 passed
- 10 failed

Let's assume **"Failed"**  as postive class. we are trying to check if that student fail or not

### Actual vs Predicted

|                | Predicted Failed | Predicted Passed |
|----------------|------------------|------------------|
| Actual Failed  | 8                | 2                |
| Actual Passed  | 5                | 85               |

### Understanding the Terms

- **True Positive (TP)** = 8  
  - Failed students correctly predicted as failed.

- **False Positive (FP)** = 5  
  - Passed students incorrectly predicted as failed.

- **False Negative (FN)** = 2  
  - Failed students incorrectly predicted as passed.

- **True Negative (TN)** = 85  
  - Passed students correctly predicted as passed.

---

## Precision

**Out of all students that our model predicted as failed, how many were actually failed?** <br>
**Hamro model le fail prediction gareko madhye kati chai real mai fail thiye?**

Formula:

Precision = TP / (TP + FP)

Calculation:

Precision = 8 / (8 + 5)

Precision = 8 / 13

Precision = 0.615 = 61.5%

### Intuition

Higher precision means fewer false alarms.

---

## Recall

**Out of all the actual failed students, how many were correctly identified as failed by our model?** <br>
**Vastav mai fail bhayeka students madhye, kati jana lai hamro model le fail bhani sahi predict garyo?**

Formula:

Recall = TP / (TP + FN)

Calculation:

Recall = 8 / (8 + 2)

Recall = 8 / 10

Recall = 0.8 = 80%

### Intuition

Recall measures how many actual failed students the model was able to catch.

Higher recall means fewer failed students are missed.

---

## F1 Score

Precision and Recall often trade off against each other.

The F1 Score combines both into a single metric.

Formula:

F1 Score = 2 × (Precision × Recall) / (Precision + Recall)

Calculation:

F1 Score = 2 × (0.615 × 0.8) / (0.615 + 0.8)

F1 Score = 0.695 = 69.5%

### Intuition

The F1 Score is high only when both Precision and Recall are high.

---

# When to Prioritize Recall vs Precision

## Prioritize Recall

Use **Recall** when missing a positive case is very costly.

**Question Recall Answers:**

> Out of all actual positive cases, how many did we correctly identify?

### Example 1: Cancer Detection

Suppose 100 patients have cancer.

* Model identifies 95 of them.
* Model misses 5 patients.

**Recall = 95%**

**वास्तवमै क्यान्सर भएका बिरामीमध्ये, कति जनालाई हाम्रो मोडेलले क्यान्सर छ भनेर सही predict गर्यो?**

This is important because missing a cancer patient could delay treatment and have serious consequences.

### Example 2: Student Failure Prediction

Suppose 20 students are actually going to fail.

* Model correctly identifies 18 students.
* Model misses 2 students.

**Recall = 90%**

**वास्तवमै फेल भएका विद्यार्थीमध्ये, कति जनालाई हाम्रो मोडेलले फेल भनेर सही predict गर्यो?**

This is important because we want to provide support to as many struggling students as possible.

**High Recall = Few False Negatives**
(हाम्रो case मा positive भनेको **Cancer छ** वा **Fail छ** हो)

* Cancer थियो तर **Cancer छैन** भन्यो.

  * Issue: Patient को treatment ढिलो हुन सक्छ।

* Fail हुने विद्यार्थी थियो तर **Pass** भन्यो.

  * Issue: Student ले आवश्यक support वा extra preparation पाउँदैन।

---

## Prioritize Precision

Use **Precision** when false alarms are costly.

**Question Precision Answers:**

> Out of all cases predicted as positive, how many were actually positive?

### Example 1: Spam Email Detection

Suppose the model marks 100 emails as spam.

* 95 are actually spam.
* 5 are important emails.

**Precision = 95%**

**हाम्रो मोडेलले Spam prediction गरेका इमेलहरूमध्ये, कति वटा वास्तवमै Spam थिए?**

High precision is important because we do not want important emails incorrectly sent to the spam folder.

**High Precision = Few False Positives**
(हाम्रो case मा positive भनेको **Spam Email** हो)

* Email Spam थिएन तर मोडेलले **Spam हो** भन्यो.

  * Issue: Important email छुट्न सक्छ।

---


### Why Recall is More Important in Cancer Detection?

In cancer detection, we usually prioritize **Recall** over **Precision**.

Why?

Because missing a patient who actually has cancer can be very dangerous.

- Cancer थियो तर मोडेलले **Cancer छैन** भन्यो.
  - This is a False Negative.
  - The patient may not receive treatment on time.
  - In severe cases, this can be life-threatening.

On the other hand:

- Cancer थिएन तर मोडेलले **Cancer छ** भन्यो.
  - This is a False Positive.
  - The patient may undergo additional tests.
  - It may cause stress and extra cost, but the patient can still be checked and confirmed by doctors.

Therefore, in cancer detection:

> It is usually better to mistakenly flag a healthy person for further testing than to miss a patient who actually has cancer.

That's why we prioritize **High Recall**.

## Should We Always Focus on Recall or Precision?

No.

The metric we prioritize depends on the business problem and the cost of mistakes.

### If False Negatives are More Dangerous

Prioritize **Recall**.

A False Negative means:

> The actual positive case exists, but the model fails to identify it.

Examples:

- Cancer detection
- Fraud detection
- Student failure prediction

In these cases, missing a positive case can have serious consequences.

---

### If False Positives are More Costly

Prioritize **Precision**.

A False Positive means:

> The model predicts positive, but the actual case is negative.

Examples:

- Spam email detection
- Candidate shortlisting
- Loan approval systems

In these cases, false alarms can waste time, money, and resources.

---

### Rule of Thumb

Ask yourself:

> Which mistake hurts more?

- If missing a positive case hurts more → Focus on **Recall**
- If false alarms hurt more → Focus on **Precision**
- If both matter → Use **F1 Score** to balance Precision and Recall

So before evaluating a model, we should first understand the business problem and the cost of different types of errors.

## Which Metric Should We Prioritize?

It depends on the goal of the system.

### If the goal is Health Screening

Prioritize **Recall** for the classes of interest (Overweight and Obese).

Why?

Because we do not want to miss people who may be overweight or obese.

Example:

- Person is actually Obese.
- Model predicts Normal Weight.

This person may not receive health advice or further screening.

In this case:

> Missing a risky patient is worse than raising a few extra alarms.

Therefore, Recall is important.

---

### If the goal is Providing Expensive Treatment or Programs

Prioritize **Precision**.

Example:

- Person is actually Normal Weight.
- Model predicts Obese.

The person may be unnecessarily enrolled in a costly health program.

In this case:

> False alarms are expensive.

Therefore, Precision is important.

---

### For BMI Classification

In most practical situations:

> Recall for Overweight and Obese classes is usually more important than Precision.

Because it is generally better to identify people who may be at health risk and then perform additional checks.

---
